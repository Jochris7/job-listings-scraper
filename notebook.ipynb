```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://realpython.github.io/fake-jobs/"
reponse = requests.get(url)
```

```python
print(reponse.status_code)
```

```python
soup = BeautifulSoup(reponse.text,'html.parser')
```

```python
cards = soup.find_all('div', class_="card")
offres = []
```

```python
for card in cards :
    titre_tag = card.find('h2', class_="title is-5")
    titre_emploi = titre_tag.text.strip() if titre_tag else "N/A"

    entreprise_tag = card.find('h3', class_="subtitle is-6 company")
    nom_entreprise = entreprise_tag.text.strip() if entreprise_tag else "N/A"

    lieu_tag = card.find('p', class_="location")
    emplacement = lieu_tag.text.strip() if lieu_tag else "N/A"

    liens = card.find_all('a', class_="card-footer-item")
    url_details = liens[1]['href'] if len(liens) > 1 else "N/A"

    offres.append({
        "titre_emploi": titre_emploi,
        "nom_entreprise": nom_entreprise,
        "emplacement": emplacement,
        "url_details": url_details
    })
```

```python
print(f"{len(offres)} offres enregistrées dans offres_emploi.csv")
```

```python
print(offres)
```

```python
import csv

with open("offres_emploi.csv", "w", newline="", encoding="utf-8") as fichier_csv:
    champs = ["titre_emploi", "nom_entreprise", "emplacement", "url_details"]
    writer = csv.DictWriter(fichier_csv, fieldnames=champs)
    writer.writeheader()
    writer.writerows(offres)

print(f"{len(offres)} offres enregistrées dans offres_emploi.csv")
```
